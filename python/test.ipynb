Example of calling a U3lib function `rhomax` defined in `python/rhomax.cpp` source file using **pybind11**. This function calculates $\rho_{\max}$ for 3 input U(3) irreps $[n_{13},n_{23},n_{33}]$, where these labels are provided as function arguments.

Exemplary code for:

$$
[4,2,0]\times[2,1,0]\to[4,2,0]
$$

where $\rho_{\max}=2$ and the number of reduced U3U2 coefficients is 49.

```python
import u3lib

#help(u3lib)
```

```python
#help(u3lib.rhomax)

u3a = [4, 2, 0]
u3b = [2, 1, 0]
u3c = [4, 2, 0]

u3lib.rhomax(*u3a, *u3b, *u3c)
```

Example of calculating reduced U3U2 coefficients by using `u3::WigU3U2_coeffs` class for:

$$
[4,2,0]\times[2,1,0]\to[4,2,0]
$$

```python
coeffs = u3lib.WigU3U2_coeffs(*u3a, *u3b, *u3c)
coeffs.size()
```
